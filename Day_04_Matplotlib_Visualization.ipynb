## Day 4 – Matplotlib Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("/content/HR-Employee-Attrition.xlsx")
```

## 1. Line chart – average MonthlyIncome by JobLevel


```python
income_by_level = df.groupby("JobLevel")["MonthlyIncome"].mean()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(income_by_level.index, income_by_level.values, marker="o")
plt.title("Average Monthly Income by Job Level")
plt.xlabel("Job Level")
plt.ylabel("Average Monthly Income")
plt.grid(True)
plt.show()
```

## 2. Scatter plot – Age vs MonthlyIncome


```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Age"], df["MonthlyIncome"], alpha=0.5)
plt.title("Age vs Monthly Income")
plt.xlabel("Age")
plt.ylabel("Monthly Income")
plt.show()
```

## 3. Bar chart – employee count by Department


```python
department_count = df["Department"].value_counts()
```

```python
plt.figure(figsize=(8, 5))
plt.bar(department_count.index, department_count.values)
plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.xticks(rotation=20)
plt.show()
```

## 4. Pie chart – Attrition distribution


```python
attrition_count = df["Attrition"].value_counts()
```

```python
plt.figure(figsize=(6, 6))
plt.pie(
    attrition_count.values,
    labels=attrition_count.index,
    autopct="%1.1f%%"
)
plt.title("Employee Attrition Distribution")
plt.show()
```

## 5. Histogram – Age distribution


```python
plt.figure(figsize=(8, 5))
plt.hist(df["Age"], bins=15, edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

## 6. Box plot – MonthlyIncome


```python
plt.figure(figsize=(8, 5))
plt.boxplot(df["MonthlyIncome"].dropna())
plt.title("Monthly Income Distribution and Outliers")
plt.ylabel("Monthly Income")
plt.show()
```
