## Day 1 – NumPy Arrays & Operations
Dataset: HR Employee Attrition
Goal: Practice NumPy arrays and basic numerical exploration.


```python

import numpy as np
import pandas as pd
```

## Load the HR dataset


```python
df = pd.read_excel("/content/HR-Employee-Attrition.xlsx")

print("Dataset shape:", df.shape)
```

## Select numerical columns


```python
numeric_data = df.select_dtypes(include=np.number)
```

## 1D array


```python
age_1d = df["Age"].to_numpy()
print(age_1d[:10])
```

## 2D array

```python
age_income_2d = df[["Age", "MonthlyIncome"]].to_numpy()
print(age_income_2d[:5])
```

## 3D array


## Create 3 layers using Age, MonthlyIncome and TotalWorkingYears.


```python
first_10 = df[["Age", "MonthlyIncome", "TotalWorkingYears"]].to_numpy()[:10]
array_3d = np.stack([first_10, first_10, first_10])
print("\n3D Array shape:", array_3d.shape)
```

## Addition, subtraction, multiplication and division


```python
a = np.array([10, 20, 30, 40, 50])
b = np.array([2, 4, 5, 8, 10])
```

```python
print("\nAddition:", a + b)
```

```python
print("Subtraction:", a - b)
```

```python
print("Multiplication:", a * b)
```

```python
print("Division:", a / b)
```

## NumPy functions


```python
print("\nMean Age:", np.mean(age_1d))
```

```python
print("Total Age:", np.sum(age_1d))
```

```python
print("Age Standard Deviation:", np.std(age_1d))
```

## reshape()


```python
sample = np.arange(1, 13)
```

```python
reshaped = sample.reshape(3, 4)
print(reshaped)
```

## arange()


```python
numbers = np.arange(0, 21, 5)
print("\narange():", numbers)
```

## linspace()


```python
line_numbers = np.linspace(0, 100, 6)
print("linspace():", line_numbers)
```

## Find indices of maximum and minimum Age


```python
max_index = np.argmax(age_1d)
min_index = np.argmin(age_1d)
```

```python
print("\nIndex of maximum Age:", max_index)
```

```python
print("Maximum Age:", age_1d[max_index])
```

```python
print("Index of minimum Age:", min_index)
```

```python
print("Minimum Age:", age_1d[min_index])
```
