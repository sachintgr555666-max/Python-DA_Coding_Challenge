```python
# Day 8 - Employee Hours Report
# Scenario: Student Exam Scores Performance Report

import random

scores = [random.randint(0, 100) for _ in range(10)]

print("--- Exam Scores ---")
print(scores)
```

```python
print("\n--- Scores with Index ---")
for index, score in enumerate(scores):
    print(f"Index {index}: Score {score}")
```

```python
high_scores = [score for score in scores if score > 75]

print("\nScores greater than 75:")
print(high_scores)
```

```python
average_score = sum(scores) / len(scores)

print(f"\nAverage Score: {average_score:.2f}")
```
