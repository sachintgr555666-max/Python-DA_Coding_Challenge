## Detect missing values


```python
print(df.isnull().sum())
```

## Create a small practice dataset with missing values


```python
practice = df[["EmployeeNumber", "MonthlyIncome"]].head(10).copy()
```

```python
practice.loc[0, "MonthlyIncome"] = None
```

```python
print(practice)
```

## fillna() – fill missing value with median


```python
practice["MonthlyIncome"] = practice["MonthlyIncome"].fillna(
    practice["MonthlyIncome"].median()
)
```

```python
print(practice)
```

## dropna() – remove rows containing missing values


```python
dropna_example = practice.copy()
```

```python
dropna_example.loc[1, "MonthlyIncome"] = None
```

```python
dropna_example = dropna_example.dropna()
```

```python
print(dropna_example)
```

## Merge two DataFrames


```python
employee_info = df[["EmployeeNumber", "Age", "Gender"]].head(10)
```

```python
income_info = df[["EmployeeNumber", "MonthlyIncome"]].head(10)
```

```python
merged_df = pd.merge(employee_info, income_info, on="EmployeeNumber", how="inner")
```

```python
merged_df.head()
```

## Concatenate two DataFrames


```python
part1 = df.head(5)
```

```python
part2 = df.tail(5)
```

```python
concat_df = pd.concat([part1, part2], ignore_index=True)
```

```python
print(concat_df[["EmployeeNumber", "Age", "Attrition"]])
```

## Join two DataFrames using EmployeeNumber as index


```python
left = df[["EmployeeNumber", "Age"]].head(10).set_index("EmployeeNumber")
```

```python
right = df[["EmployeeNumber", "MonthlyIncome"]].head(10).set_index("EmployeeNumber")
```

```python
joined_df = left.join(right, how="inner")
```

```python
print(joined_df.head())
```

## Group by and aggregate


```python
department_summary = df.groupby("Department").agg(
    Average_Income=("MonthlyIncome", "mean"),
    Average_Age=("Age", "mean"),
    Employee_Count=("EmployeeNumber", "count")
).reset_index()
```

```python
print(department_summary)
```

## Another useful grouping


```python
attrition_summary = df.groupby("Attrition").agg(
    Average_Income=("MonthlyIncome", "mean"),
    Average_Age=("Age", "mean"),
    Employee_Count=("EmployeeNumber", "count")
).reset_index()
```

```python
print(attrition_summary)
```
