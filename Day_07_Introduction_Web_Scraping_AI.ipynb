# Day 7 – Introduction to Web Scraping & AI

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Fetch HTML content
url = "https://books.toscrape.com/"
response = requests.get(url, timeout=15)
```

```python
print("Status code:", response.status_code)
```

```python
print("Content type:", response.headers.get("content-type"))
```

```python
html_content = response.text
```

```python
print("HTML characters:", len(html_content))
```

```python
print(html_content[:500])
```

## AI-assisted code generation with Gemini in Colab

```python
soup = BeautifulSoup(html_content, "html.parser")

books = []

for article in soup.select("article.product_pod"):
    title_tag = article.select_one("h3 a")
    price_tag = article.select_one(".price_color")

    if title_tag and price_tag:
        books.append({
            "BookTitle": title_tag.get("title", title_tag.get_text(strip=True)),
            "PriceText": price_tag.get_text(strip=True)
        })

scraped_books = pd.DataFrame(books)

print(scraped_books.head())
print("\nRows scraped:", len(scraped_books))
```

```python
scraped_books["Price"] = (
    scraped_books["PriceText"].str.replace("Â", "", regex=False).str.replace("£", "", regex=False).astype(float)
)
```

```python
scraped_books.drop(columns=["PriceText"], inplace=True)
```
