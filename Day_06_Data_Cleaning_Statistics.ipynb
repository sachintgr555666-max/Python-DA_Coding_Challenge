## Day 6 – Data Cleaning & Statistics

```python
import pandas as pd
```

## Load dataset


```python
df = pd.read_excel("/content/HR-Employee-Attrition.xlsx")
```

```python
df.shape
```

## 1. Detect duplicates


```python
duplicate_count = df.duplicated().sum()
print("\nNumber of duplicate rows:", duplicate_count)
```

## Remove duplicates


```python
df = df.drop_duplicates()
```

```python
print("Shape after removing duplicates:", df.shape)
```

## 2. Detect missing data


```python
print(df.isnull().sum())
```

## Handle missing numerical values using median


```python
numeric_columns = df.select_dtypes(include="number").columns
```

```python
for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())
```

## Handle missing categorical values using mode


```python
categorical_columns = df.select_dtypes(exclude="number").columns
```

```python
for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print(df.isnull().sum())
```

## 3. Basic statistics


```python
print(df[numeric_columns].mean())
```

```python
print(df[numeric_columns].median())
```

```python
print(df[numeric_columns].mode().iloc[0])
```

```python
print(df[numeric_columns].std())
```

```python
print(df[numeric_columns].var())
```

## 4. Group by categories and summarize insights


```python
department_summary = df.groupby("Department").agg(
    Employee_Count=("EmployeeNumber", "count"),
    Average_Age=("Age", "mean"),
    Average_Income=("MonthlyIncome", "mean"),
    Average_Job_Satisfaction=("JobSatisfaction", "mean")
).reset_index()
```

```python
print(department_summary)
```

## Attrition summary


```python
attrition_summary = df.groupby("Attrition").agg(
    Employee_Count=("EmployeeNumber", "count"),
    Average_Age=("Age", "mean"),
    Average_Income=("MonthlyIncome", "mean"),
    Average_Job_Satisfaction=("JobSatisfaction", "mean")
).reset_index()
```

```python
print(attrition_summary)
```

## Save the cleaned dataset


```python
df.to_csv("HR_Employee_Attrition_Cleaned.csv", index=False)
```
